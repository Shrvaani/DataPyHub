```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../data/processed/cleaned_ipl.csv")

plt.figure(figsize=(10,5))
sns.countplot(data=df, x='season')
plt.title("Matches Played per Season")
plt.xticks(rotation=45)
plt.show()

team_wins = df['winner'].value_counts().head(10)
team_wins.plot(kind='bar', color='skyblue', figsize=(10,5))
plt.title("Top Teams by Wins")
plt.show()

sns.histplot(df['win_by_runs'], bins=20, color='orange', kde=True)
plt.title("Distribution of Win Margins (Runs)")
plt.show()
```
